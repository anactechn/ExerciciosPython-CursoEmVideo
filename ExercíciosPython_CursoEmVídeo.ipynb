```python
#Ex.98-Curso de Python 3 - Mundo 3

from time import sleep

def contador_brincadeira(inicio, fim, passo):
    print(f"\n🔢 Vamos contar de {inicio} até {fim}, pulando de {abs(passo)} em {abs(passo)}!")
    sleep(1)

    if passo == 0:
        passo = 1
        print("🧙‍♂️ Opa! Bruxo travou, vou consertar o passo para 1!")

    if inicio < fim:
        print("⬆️ Contagem crescente! Preparar...")
        passo = abs(passo)
    else:
        print("⬇️ Contagem decrescente! Atenção...")
        passo = -abs(passo)

    sleep(1)
    print("🎉 Começando!")

    for num in range(inicio, fim + (1 if inicio < fim else -1), passo):
        print(num, end=' ', flush=True)
        sleep(0.3)
        if num % 5 == 0:
            print("✨", end=' ')

    print("\n🏁 FIM DA CORRIDA NUMÉRICA!")
    print("—" * 40)
    sleep(1)

    if abs(fim - inicio) > 20:
        print("😱 Nossa! Que contagem longa!")
    else:
        print("😊 Foi divertido contar com você!")

print("🌟 BEM-VINDO AO JOGO DA CONTAGEM MÁGICA! 🌟")
sleep(1)

print("\nPrimeira rodada mágica:")
contador_brincadeira(1, 10, 1)

print("\nSegunda rodada emocionante:")
contador_brincadeira(10, 0, 2)

print("\nAgora é SUA vez de criar magia!")
contador_brincadeira(
    int(input("Digite o número de início: ")),
    int(input("Digite o número final: ")),
    int(input("Digite o tamanho do pulo: "))
)

print("\n🎊 Parabéns! Você dominou a magia dos números! 🎊")
```

```python
#Ex.99-Curso de Python 3 - Mundo 3

from time import sleep

def super_detetive(*pistas):
    print("\n🕵️‍♂️ Bem-vindo ao Jogo do Super Detetive Numérico!")
    sleep(1)

    if not pistas:
        print("🔍 Opa! Não encontrei nenhuma pista...")
        print("🏆 Mas você ganhou um abraço virtual! 🤗")
        return 0

    print(f"\n🔎 Analisando as pistas: {pistas}")
    sleep(1)

    print("\n💡 Processando...")
    sleep(2)

    maior_pista = max(pistas)

    print(f"\n🎉 EU DESCOBRI! O maior número é... {maior_pista}!")
    sleep(1)

    if maior_pista > 50:
        print("🤯 UAU! Esse número é GIGANTESCO!")
    elif maior_pista > 10:
        print("😃 Boa pista! Número bem grandinho!")
    else:
        print("🥰 Que fofurinha! Um número pequenino!")

    return maior_pista

print("🌟 Vamos brincar de detetive de números? 🌟")
sleep(1)

print("\nPrimeira investigação:")
super_detetive(3, 7, 2, 8, 5)

print("\nSegunda investigação:")
super_detetive(10, 20, 30)

print("\nTerceira investigação:")
super_detetive()

print("\nAgora é sua vez de investigar!")
numeros = input("Digite os números separados por vírgula: ").split(',')
numeros = [int(num.strip()) for num in numeros]
super_detetive(*numeros)

print("\n🏅 Parabéns, Super Detetive Numérico! 🏅")
```

```python
#Ex.100-Curso de Python 3 - Mundo 3

from random import randint
from time import sleep

def sorteio_magico():
    print("🎩✨ Vamos começar o Sorteio Mágico de Números! ✨🎩")
    sleep(1)

    numeros_sorteados = []
    print("\n🧙‍♂️ O Mago está sorteando 5 números mágicos...")

    for _ in range(5):
        num = randint(1, 10)
        numeros_sorteados.append(num)
        print(f"⚡ {num}", end=' ', flush=True)
        sleep(0.7)

    print("\n\n🎉 PRONTINHO! Os números mágicos são:", numeros_sorteados)
    return numeros_sorteados

def soma_encantada(numeros):
    print("\n🔮 Agora vamos somar os números pares mágicos...")
    sleep(1)

    pares = [num for num in numeros if num % 2 == 0]
    total = sum(pares)

    print(f"\n✨ Números pares encontrados: {pares}")
    print(f"💫 A soma mágica é: {total}")

    if total > 10:
        print("😲 UAU! Que soma poderosa!")
    elif total > 0:
        print("😊 Boa energia mágica!")
    else:
        print("🧐 Oops... nenhum número par encontrado!")

    return total

# Programa principal
print("🌟 BEM-VINDO AO MUNDO DA MATEMÁGICA! 🌟")
sleep(1)

numeros_magicos = sorteio_magico()
soma_encantada(numeros_magicos)

print("\n🌈 Obrigado por jogar! Até a próxima aventura mágica! 🌈")
```

```python
#Ex.101-Curso de Python 3 - Mundo 3

from datetime import datetime

def verifica_voto(ano_nascimento):
    idade = datetime.now().year - ano_nascimento

    if idade < 16:
        return "🛑 VOTO NEGADO"
    elif 16 <= idade < 18 or idade >= 70:
        return "🟡 VOTO OPCIONAL"
    else:
        return "🟢 VOTO OBRIGATÓRIO"

# Programa principal
print("🗳️  VERIFICADOR DE SITUAÇÃO ELEITORAL  🗳️")
print("--------------------------------------")

ano = int(input("Digite seu ano de nascimento: "))
resultado = verifica_voto(ano)

print("\n🔍 Analisando sua situação...")
print(f"\nSua idade: {datetime.now().year - ano} anos")
print(f"Situação eleitoral: {resultado}")

if "NEGADO" in resultado:
    print("\n📢 Você ainda não pode votar! Aproveite para estudar!")
elif "OPCIONAL" in resultado:
    print("\n📢 Você pode votar se quiser! É seu direito!")
else:
    print("\n📢 Não esqueça de votar! É seu dever cívico!")
```

```python
#Ex.102-Curso de Python 3 - Mundo 3

from math import factorial
from time import sleep

def super_fatorial(n, show=False):
    """
    Calcula o fatorial de um número com modo show opcional.
    :param n: Número a ser calculado
    :param show: (opcional) Mostrar ou não o cálculo
    :return: O valor do fatorial de n
    """
    if show:
        print(f"\n🔢 Vamos calcular {n}! passo a passo:")
        sleep(1)
        resultado = 1
        for i in range(n, 0, -1):
            resultado *= i
            print(f"{i} {'×' if i > 1 else '='} ", end='', flush=True)
            sleep(0.3)
        print(f"{resultado}")
    else:
        resultado = factorial(n)

    return resultado

# Programa principal
print("🌟 CALCULADORA DE FATORIAL MÁGICA 🌟")
print("-----------------------------------")

num = int(input("Digite um número para calcular o fatorial: "))
mostrar = input("Quer ver o cálculo passo a passo? [S/N]: ").strip().upper()[0] == 'S'

resultado = super_fatorial(num, mostrar)

print(f"\n🎉 O fatorial de {num} é: {resultado}")

if resultado > 1000000:
    print("🤯 Esse número é ENORME!")
elif num >= 10:
    print("😮 Que cálculo impressionante!")
else:
    print("😊 Fatorial calculado com sucesso!")
```

```python
#Ex.104-Curso de Python 3 - Mundo 3

def leia_int(mensagem):
    """
    Função que funciona como input(), mas só aceita valores numéricos inteiros
    :param mensagem: Mensagem que será exibida para o usuário
    :return: Valor numérico inteiro digitado pelo usuário
    """
    while True:
        entrada = input(mensagem).strip()

        if entrada.isdigit():
            return int(entrada)
        else:
            print("\033[31mERRO! Digite um número inteiro válido.\033[m")

# Exemplo de uso
print("🧮 VALIDADOR DE NÚMEROS INTEIROS 🧮")
print("----------------------------------")
numero = leia_int("Digite um número inteiro: ")
print(f"✅ Você digitou o número: {numero}")
```

```python
#Ex.105-Curso de Python 3 - Mundo 3

from statistics import mode, median

# Definindo cores para formatação
CORES = {
    'Boa': '\033[1;32m',      # Verde
    'Razoável': '\033[1;33m',  # Amarelo
    'Ruim': '\033[1;31m',      # Vermelho
    'reset': '\033[0m'         # Reset
}

def notas(*notas, situação=False, grafico=False):
    """
    Analisa as notas de alunos e retorna um dicionário com estatísticas.

    Parâmetros:
        *notas (float): Uma ou mais notas dos alunos (aceita várias)
        situação (bool, opcional): Se True, inclui a situação da turma
        grafico (bool, opcional): Se True, exibe um gráfico ASCII simples

    Retorna:
        dict: Dicionário com estatísticas completas da turma
    """
    if not notas:
        return {}

    estatisticas = {
        'total': len(notas),
        'maior': max(notas),
        'menor': min(notas),
        'média': sum(notas)/len(notas),
        'moda': mode(notas),
        'mediana': median(sorted(notas))
    }

    if situação:
        media = estatisticas['média']
        if media >= 8:
            situ = 'Boa'
        elif media >= 6:
            situ = 'Razoável'
        else:
            situ = 'Ruim'
        estatisticas['situação'] = situ

    if grafico:
        print("\n📊 GRÁFICO DE DISTRIBUIÇÃO:")
        max_count = max(notas.count(n) for n in set(notas))
        for nota in sorted(set(notas)):
            count = notas.count(nota)
            bar = '█' * int(count * 20 / max_count) if max_count > 0 else ''
            print(f"{nota:4.1f} | {bar} ({count})")

    return estatisticas

# Programa principal
print("📊 ANALISADOR DE NOTAS DA TURMA 📊")
print("---------------------------------")

notas_alunos = []
while True:
    entrada = input("Digite uma nota (0-10) ou 's' para sair: ").strip().lower()
    if entrada == 's':
        break
    try:
        nota = float(entrada)
        if 0 <= nota <= 10:
            notas_alunos.append(nota)
        else:
            print("⚠️ Nota deve estar entre 0 e 10!")
    except ValueError:
        print("⚠️ Valor inválido! Digite apenas números.")

if notas_alunos:
    resultado = notas(*notas_alunos, situação=True, grafico=True)

    print("\n📈 RESULTADO FINAL:")
    for chave, valor in resultado.items():
        if chave == 'situação':
            cor = CORES.get(valor, '')
            print(f"{chave.upper()}: {cor}{valor}{CORES['reset']}")
        else:
            print(f"{chave.upper()}: {valor:.2f}" if isinstance(valor, float) else f"{chave.upper()}: {valor}")
else:
    print("\nNenhuma nota válida foi informada.")
```

```python
#Ex.106-Curso de Python 3 - Mundo 3

from time import sleep

# Códigos de cores ANSI
cores = {
    'titulo': '\033[1;34m',    # Azul brilhante
    'texto': '\033[1;37m',     # Branco
    'comando': '\033[1;32m',   # Verde
    'fim': '\033[0m',          # Reset
    'destaque': '\033[1;33m',  # Amarelo
    'erro': '\033[1;31m'       # Vermelho
}

def ajuda(comando):
    """
    Exibe o help personalizado com cores para qualquer comando Python
    """
    print(f"{cores['titulo']}⋆｡ﾟ✶°。⋆｡ﾟ✶°。⋆｡ﾟ✶°。⋆")
    print(f"{cores['destaque']}  Acessando o manual do comando '{comando}'")
    print(f"{cores['titulo']}⋆｡ﾟ✶°。⋆｡ﾟ✶°。⋆｡ﾟ✶°。⋆{cores['fim']}")
    sleep(1)
    print(f"{cores['texto']}")
    help(comando)
    print(f"{cores['fim']}")

# Programa principal
print(f"{cores['titulo']}↭✧.SISTEMA DE AJUDA INTERATIVO.✧↭{cores['fim']}")
print(f"{cores['comando']}Digite 'FIM' para sair{cores['fim']}")

while True:
    print(f"\n{cores['destaque']}~" * 30)
    cmd = input(f"{cores['comando']}Função ou Biblioteca > {cores['fim']}").strip().lower()

    if cmd == 'fim':
        print(f"{cores['titulo']}✧･ﾟ: *✧･ﾟ:* ATÉ LOGO! *:･ﾟ✧*:･ﾟ✧{cores['fim']}")
        break
    elif cmd:
        ajuda(cmd)
    else:
        print(f"{cores['erro']}ERRO! Digite um comando válido.{cores['fim']}")
```
